The objective of this task is to implement Linear Regression and Logistic Regression from scratch using Gradient Descent without using machine learning libraries. The models are trained on real-world datasets and evaluated using standard performance metrics.

```python
import numpy as np
import pandas as pd

from linear_regression import LinearRegressionScratch
from logistic_regression import LogisticRegressionScratch
```

# Part 1 - Linear Regression

```python
# Load regression dataset

oil_df = pd.read_csv("../data/oil_sales_assignment_dataset.csv")

oil_df.head()
```

```python
# Prepare features and target

X = oil_df[
    [
        "year",
        "month",
        "volume_sales",
        "average_price"
    ]
].values.astype(float)

X = (X - X.mean(axis=0)) / X.std(axis=0)

y = oil_df["value_sales"].values.astype(float)

print(X.shape)
print(y.shape)
```

```python
# Train Linear Regression

linear_model = LinearRegressionScratch(
    learning_rate=0.001,
    iterations=3000
)

linear_model.fit(X, y)
```

```python
predictions = linear_model.predict(X)

print("MSE :", linear_model.mean_squared_error(y, predictions))
print("RMSE:", linear_model.root_mean_squared_error(y, predictions))
print("R² :", linear_model.r2_score(y, predictions))
```

```python
results = pd.DataFrame({
    "Actual": y,
    "Predicted": predictions
})

results.head()
```

# Part 2 - Logistic Regression

```python
heart_df = pd.read_csv("../data/heart_disease_risk_2026.csv")

heart_df.head()
```

```python
feature_columns = [
    "age",
    "resting_bp_systolic",
    "resting_bp_diastolic",
    "cholesterol_total",
    "hdl",
    "ldl",
    "triglycerides",
    "fasting_blood_sugar",
    "hba1c",
    "bmi",
    "resting_heart_rate",
    "max_heart_rate_achieved",
    "st_depression",
    "alcohol_units_per_week",
    "exercise_minutes_per_week",
    "sleep_hours",
    "stress_score",
    "daily_steps",
    "diet_quality_score"
]

X = heart_df[feature_columns].values.astype(float)

X = (X - X.mean(axis=0)) / X.std(axis=0)

y = heart_df["has_heart_disease"].values.astype(int)

print(X.shape)
print(y.shape)
```

```python
logistic_model = LogisticRegressionScratch(
    learning_rate=0.01,
    iterations=5000
)

logistic_model.fit(X, y)
```

```python
predictions = logistic_model.predict(X)

accuracy = logistic_model.accuracy(y, predictions)

print("Accuracy:", accuracy)
```

```python
results = pd.DataFrame({
    "Actual": y,
    "Predicted": predictions
})

results.head()
```
